```python
# PARCIAL 2 - INVERNADERO EN OBSERVACIÓN
# Andrés Felipe Pabón Barbosa

import csv

tempOptima = (18, 26)
tempCritica = (12, 32)
tempSensor = (-10, 60)
humOptima = (60, 80)
humCritica = (40, 90)
humSensor = (0, 100)

archivo = "invernadero.csv"
lecturas = []

# IMPORTAR CSV
try:
    with open(archivo, "r", encoding="utf-8") as file:
        lecturas = list(csv.DictReader(file))
except:
    pass

opcion = ""

while opcion != "0":

    print("\n1. Agregar")
    print("2. Leer")
    print("3. Actualizar")
    print("4. Borrar")
    print("5. Reporte")
    print("0. Salir")

    opcion = input("Opción: ")

    # AGREGAR
    if opcion == "1":

        while len(lecturas) < 8:

            tempTexto = input("Temperatura: ").replace(",", ".")
            humTexto = input("Humedad: ").replace(",", ".")

            try:
                temp = float(tempTexto)
                hum = float(humTexto)

                if tempSensor[0] <= temp <= tempSensor[1] and humSensor[0] <= hum <= humSensor[1]:
                    break

            except:
                pass

            print("Datos inválidos.")

        if len(lecturas) < 8:

            if not tempCritica[0] <= temp <= tempCritica[1] or not humCritica[0] <= hum <= humCritica[1]:
                estado = "Crítico"
                accion = "Activar protocolo de emergencia"

            elif tempOptima[0] <= temp <= tempOptima[1] and humOptima[0] <= hum <= humOptima[1]:
                estado = "Óptimo"
                accion = "Condiciones óptimas"

            else:
                estado = "Observación"
                accion = "Ventilar" if temp > 26 else "Calefacción" if temp < 18 else "Deshumidificar" if hum > 80 else "Nebulizar" if hum < 60 else "Mantener"

            lecturas.append({
                "lectura": len(lecturas) + 1,
                "temperatura": temp,
                "humedad": hum,
                "estado": estado,
                "accion": accion
            })

    # LEER
    elif opcion == "2":

        i = 0

        while i < len(lecturas):
            print(
                lecturas[i]["lectura"],
                lecturas[i]["temperatura"], "°C",
                lecturas[i]["humedad"], "%",
                lecturas[i]["estado"],
                lecturas[i]["accion"]
            )
            i += 1

    # ACTUALIZAR
    elif opcion == "3":

        posicion = int(input("Lectura: ")) - 1

        if 0 <= posicion < len(lecturas):

            temp = float(input("Nueva temperatura: ").replace(",", "."))
            hum = float(input("Nueva humedad: ").replace(",", "."))

            if tempSensor[0] <= temp <= tempSensor[1] and humSensor[0] <= hum <= humSensor[1]:

                if not tempCritica[0] <= temp <= tempCritica[1] or not humCritica[0] <= hum <= humCritica[1]:
                    estado = "Crítico"
                    accion = "Activar protocolo de emergencia"

                elif tempOptima[0] <= temp <= tempOptima[1] and humOptima[0] <= hum <= humOptima[1]:
                    estado = "Óptimo"
                    accion = "Condiciones óptimas"

                else:
                    estado = "Observación"
                    accion = "Ventilar" if temp > 26 else "Calefacción" if temp < 18 else "Deshumidificar" if hum > 80 else "Nebulizar" if hum < 60 else "Mantener"

                lecturas[posicion]["temperatura"] = temp
                lecturas[posicion]["humedad"] = hum
                lecturas[posicion]["estado"] = estado
                lecturas[posicion]["accion"] = accion

            else:
                print("Valores fuera del rango.")

    # BORRAR
    elif opcion == "4":

        posicion = int(input("Lectura: ")) - 1

        if 0 <= posicion < len(lecturas):
            lecturas.pop(posicion)

            i = 0
            while i < len(lecturas):
                lecturas[i]["lectura"] = i + 1
                i += 1

    # REPORTE
    elif opcion == "5":

        if lecturas:

            temp = [float(x["temperatura"]) for x in lecturas]
            hum = [float(x["humedad"]) for x in lecturas]

            print("\nLecturas:", len(lecturas))
            print("Temperatura:", min(temp), max(temp), sum(temp) / len(temp))
            print("Humedad:", min(hum), max(hum), sum(hum) / len(hum))

    # GUARDAR CSV
    with open(archivo, "w", newline="", encoding="utf-8") as file:
        campos = ["lectura", "temperatura", "humedad", "estado", "accion"]
        escritor = csv.DictWriter(file, fieldnames=campos)
        escritor.writeheader()
        escritor.writerows(lecturas)

print("Programa finalizado.")
```
